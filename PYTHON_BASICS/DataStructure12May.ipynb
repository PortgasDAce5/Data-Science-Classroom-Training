#SET OPERATIONS

```python
a = {1,2,3,4,5}
b = {4,5,6,7,8}
c = {8,9,10}
```

```python
a.union(b)  #union function to combine values from one set and the unique one from other
```

```python
a.union(b,c) #more than 1 argument passing is allowed
```

```python
b.union(a,c)
```

```python
a | b    #union funtion symbol
```

```python
a | b | c   #or c|b|a can be written as well
```

```python
a.intersection(b) #intersection funtion return new set of common elements only
```

```python
a.intersection(c)  #empty list as noting is common b/w those 2 sets
```

```python
a & b   #intersection funtion symbol 
```

```python
a & b & c
```

```python
a.difference(b)   #elements in first set that are not in the second
```

```python
b.difference(a)
```

```python
a.symmetric_difference(b)   #elements that are not common in both set
```

```python
b.symmetric_difference(c)
```

```python
a - b  #difference symbol
```

```python
a - b - c
```

```python
a ^ b #symmetric difference symbol
```

```python
a ^ b ^ c
```

```python
a.update(b)    #modifies the original set by adding elements from other
a
```

```python
a1 = {1,2,3,4,5}
b1 = {4,5,6,7,8}
c1 = {8,9,10}
```

```python
a1.intersection_update(b1)  #update the fisrt set by keeping only the common elements
a1
```

```python
b1.difference_update(c1)  #update the first set by just removing the elements found in other set
b1
```

```python
b1.symmetric_difference_update(c1)   #update the first set by keeping only the unique elements in both
b1
```

#Superset, Subset and Disjoint

```python
a2 = {1,2,3,4,5,6,7,8,9}
b2 = {3,4,5,6,7,8}
c2 = {10,20,30,40}
```

```python
a2.issuperset(b2) #checks if every element in second set is present in first
```

```python
b2.issuperset(a2)
```

```python
a2.isdisjoint(c2)  #checks if both set has nothing in common
```

```python
c2.isdisjoint(a2)
```

```python
a2.isdisjoint(b2) 
```

```python
b2.issubset(a2)  #checks if every element in first set is present in second set
```

```python
a2.issubset(b2)
```

```python
b2.isdisjoint(c2)
```

```python
a3 = {1,2,3,4,5,6,7,8,9}
b3 = {15,25,35,45}
c3 = {10,20,30,40}
```

```python
a3.issuperset(b3)
```

```python
a3.isdisjoint(b3)
```
